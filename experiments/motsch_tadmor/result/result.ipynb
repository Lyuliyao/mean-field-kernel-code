```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
data_train = np.load("../SIM3/data/x_save_1.npy")
data_test = np.load("../TEST_DATA3/data/opinion_traj_1.npy")
```

```python
def kde_pdf(x, grid, bw=None):
    """Gaussian KDE (Silverman's rule if bw is None), returns density on grid."""
    x = np.asarray(x).ravel()
    n = x.size
    if n < 2 or np.std(x) == 0:
        return np.zeros_like(grid)
    if bw is None:
        bw = 1.06 * np.std(x) * n ** (-1/5)  # Silverman's rule
        if bw <= 0 or not np.isfinite(bw):
            bw = 1.0
    u = (grid[:, None] - x[None, :]) / bw       # (G, n)
    phi = np.exp(-0.5 * u*u) / np.sqrt(2*np.pi) # standard normal pdf
    return phi.mean(axis=1) / bw                 # density integrates to 1

def style_axes(ax, show_y=True, f_size=24):
    for side in ['right','left','top','bottom']:
        ax.spines[side].set_linewidth(2)
    ax.xaxis.set_tick_params(which='major', size=6, width=2.5, direction='out', top=False)
    ax.xaxis.set_tick_params(which='minor', size=3, width=2, direction='out', top=False)
    ax.yaxis.set_tick_params(which='major', size=6, width=2.5, direction='out', right=False)
    ax.yaxis.set_tick_params(which='minor', size=3, width=2, direction='out', right=False)
    if not show_y:
        ax.yaxis.set_visible(False)
    ax.tick_params(labelsize=f_size)

# figure + panels
plt.figure(figsize=(18, 6))
plt.subplots_adjust(wspace=0, hspace=0)
f_size = 24

# common grid for KDEs
xmin = min(data_train[0,:].min(), data_test[0,:].min(), -2.5)
xmax = max(data_train[-1,:].max(), data_test[-1,:].max(),  2.5)
grid = np.linspace(xmin, xmax, 512)

# Panel 1: t=0
ax1 = plt.subplot(1,3,1)
ax1.set_title(r"$t=0$", fontsize=f_size)
style_axes(ax1, show_y=True, f_size=f_size)

# ax1.hist(data_train[0,:], bins=50, density=True, histtype="step", label="train")
# ax1.hist(data_test[0,:],  bins=50, density=True, histtype="step", label="test")

pdf_train = kde_pdf(data_train[0,:], grid)
pdf_test  = kde_pdf(data_test[0,:],  grid)
ax1.plot(grid, pdf_train, label="Mean-field Neural network")
ax1.plot(grid, pdf_test,  label="Simulation")

ax1.set_xticks([-2,0,2]); ax1.set_yticks([0,0.2,0.4])
ax1.legend(fontsize=f_size)

# Panel 2: t=2  (index 200)
ax2 = plt.subplot(1,3,2)
ax2.set_title(r"$t=2$", fontsize=f_size)
style_axes(ax2, show_y=False, f_size=f_size)

# ax2.hist(data_train[200,:], bins=50, density=True, histtype="step", label="train")
# ax2.hist(data_test[200,:],  bins=50, density=True, histtype="step", label="test")

pdf_train = kde_pdf(data_train[200,:], grid)
pdf_test  = kde_pdf(data_test[200,:],  grid)
ax2.plot(grid, pdf_train)
ax2.plot(grid, pdf_test)

ax2.set_xticks([-1,0,1])

# Panel 3: t=4  (last)
ax3 = plt.subplot(1,3,3)
ax3.set_title(r"$t=4$", fontsize=f_size)
style_axes(ax3, show_y=False, f_size=f_size)

# ax3.hist(data_train[-1,:], bins=50, density=True, histtype="step", label="train")
# ax3.hist(data_test[-1,:],  bins=50, density=True, histtype="step", label="test")

pdf_train = kde_pdf(data_train[-1,:], grid)
pdf_test  = kde_pdf(data_test[-1,:],  grid)
ax3.plot(grid, pdf_train)
ax3.plot(grid, pdf_test)

ax3.set_xticks([-1,0,1])

plt.savefig("case2_1.pdf", dpi=600, bbox_inches='tight')
```
